# Ali ke paas 500 rupees hain. Usne 150 rupees spend kiye. Variables use karke calculate karein ke Ali ke paas kitne rupees bachay

```python
total_rupee = 500
spent_rupee = 150
remaining_rupee = total_rupee - spent_rupee

print(remaining_rupee)
```

# Ek student ke total marks 450 hain aur usne 350 marks liye. Variables use karke calculate karein ke remaining marks kitne hain

```python
total_marks = 450
obtained_marks = 350
remaining_marks = total_marks - obtained_marks

print(remaining_marks)
```

# Ek person ki monthly salary 50000 hai aur expenses 32000 hain. Variables use karke calculate karein ke saving kitni hai.

```python
monthly_salary = 50000
monthly_expenses = 32000
monthly_saving = monthly_salary - monthly_expenses

print(monthly_saving)
```
